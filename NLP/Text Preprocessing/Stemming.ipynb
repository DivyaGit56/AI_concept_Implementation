#STEMMING

```python
#Stemming is an NLP technique that reduces a word to its root/base form by removing prefixes or suffixes, usually using simple rules.
```

```python
words = ["running", "runed" "jumps", "easily", "go","goes" ,"fairly"]
```

```python
###porterStemmer
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
for word in words:
    print(f"Original Word: {word} --> Stemmed Word: {stemmer.stem(word)}")
```

```python
# Disadvantages of Stemming in NLP
# Can produce invalid words
# studies → studi
# running → run ✅ but not always accurate.
# Can change the meaning
# Different words may be reduced to the same stem even when their meanings differ.
# Over-stemming
# Different meaningful words may incorrectly become the same stem.
# Example: university and universal might be reduced too aggressively by some stemmers.
# Under-stemming
# Words that should have the same root may remain different.
# Language dependent
# Stemming rules designed for English may not work well for other languages.
# Loss of linguistic information
# It removes word structure without understanding the actual meaning or grammar.
```
