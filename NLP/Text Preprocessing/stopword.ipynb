### STOPWORD

# Stop words are common words that are often removed from text during Natural Language Processing (NLP) because they usually contribute little meaning to the main topic of the text.

```python
paragraph = """Good morning everyone.

Today, I am going to talk about **Stop Words in Natural Language Processing**, or NLP.

Stop words are commonly used words that usually do not carry much important information in a sentence. Examples of stop words in English are **“the,” “is,” “am,” “are,” “a,” “an,” “in,” “on,” “and,” and “to.”**

For example, consider the sentence: **“The boy is playing in the garden.”**

If we remove the stop words, the sentence becomes **“boy playing garden.”** The important words are still present, while unnecessary common words have been removed.

The main purpose of removing stop words is to **reduce the size of the text, improve processing speed, and allow NLP systems to focus on more meaningful words**.

Stop-word removal is commonly used in tasks such as **text classification, information retrieval, and search systems**.

However, we should not always remove stop words. Some words can be important depending on the application. For example, in the sentence **“This movie is not good,”** the word **“not”** is very important because removing it changes the meaning of the sentence.

So, stop-word removal should be applied carefully according to the NLP task.

Thank you."""
```

```python
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
```

```python
import nltk
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
from nltk.stem import PorterStemmer
porter_stemmer = PorterStemmer()
```

```python
# tokenization
sentence = nltk.sent_tokenize(paragraph)
```

```python
# Apply stop word and then apply filter and stemming
for i in range(len(sentence)):
    words = nltk.word_tokenize(sentence[i])
    filtered_words = [word for word in words if word.lower() not in stopwords.words('english')]
    stemmed_words = [porter_stemmer.stem(word) for word in filtered_words]
    print("Original Sentence:", sentence[i])
    print("Filtered Words:", filtered_words)
    print("Stemmed Words:", stemmed_words)
    print("\n")
```
