# NLTK Library

NLTK (Natural Language Toolkit) is a Python library used for Natural Language Processing (NLP).

It provides tools to process and analyze human language.

Common uses of NLTK
Tokenization – breaking text into words or sentences
Stop-word removal – removing common words like is, the, a, an
Stemming – reducing words to their root form
Lemmatization – finding the meaningful base form of a word
Part-of-Speech (POS) tagging – identifying noun, verb, adjective, etc.
Named Entity Recognition (NER) – identifying people, places, organizations, etc.
Text classification – classifying text into categories

```python
# corpus  is a collection of text documents. In NLP, we use corpora to train and evaluate models.

corpus = """Hello everyone , My name is Divya and I am final year student of B.Tech in Computer Science and Engineering. I am currently working on NLP project and I am learning about Tokenization in NLP. Tokenization is the process of breaking down a text into smaller pieces called tokens. These tokens can be words, phrases, or even entire sentences. Tokenization is an important step in NLP because it allows us to analyze and understand the structure of a text. In this notebook, we will explore different methods of tokenization using Python and NLTK library."""

```

```python
corpus
```

```python
# tokenization ---> sentences
from nltk.tokenize import  sent_tokenize
```

```python
documents =sent_tokenize(corpus)
```

```python
documents
```

```python
for i in documents:
    print(i)
```

```python
# Tokenization 
# paragraph ---> words
# sentence ---> words

from nltk.tokenize import word_tokenize
```

```python
sentences = sent_tokenize(corpus)
```

```python
for sentences in documents:
    print(word_tokenize(sentences))
```
