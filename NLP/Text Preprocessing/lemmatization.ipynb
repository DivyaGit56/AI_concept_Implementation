### LEMMATIZATION

#Lemmatization is the process of reducing a word to its base or dictionary form, called a lemma.

#Lemmatization vs. stemming: Stemming simply chops off prefixes/suffixes and may produce a non-word (e.g., studies → studi), while lemmatization uses vocabulary and grammatical information to produce a valid base word (studies → study).

 

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
""" POS tagging is important for lemmatization because the same word can have different meanings and forms depending on its part of speech. For example, the word "running" can be a verb (as in "I am running") or a noun (as in "Running is fun"). 
noun-n, verb-v, adjective-a, adverb-r"""
lemmatizer.lemmatize("going", pos='v')
```

```python
word= ["going", "went", "gone", "better", "best", "worse", "worst"]
for w in word:
    print(f"{w} : {lemmatizer.lemmatize(w, pos='v')}")
```
